```python
import tensorflow
from keras.applications.resnet50 import ResNet50, preprocess_input
from keras.layers import GlobalMaxPooling2D
```

```python
model=ResNet50(weights='imagenet',include_top=False, input_shape=(224,224,3))
model.trainable=False
```

```python
model=tensorflow.keras.Sequential([
    model,
    GlobalMaxPooling2D()
])
model.summary()
```

```python
def extract_feature(img_path, model):
    img=cv2.imread(img_path)
    img=cv2.resize(img, (224,224))
    img=np.array(img)
    expand_img=np.expand_dims(img, axis=0)
    pre_img=preprocess_input(expand_img)
    result=model.predict(pre_img).flatten()
    normalized=result/norm(result)
    return normalized
```

```python
import os
from tqdm import tqdm
```

```python
filename=[]
feature_list=[]

for file in os.listdir('/content/drive/MyDrive/Colab Notebooks/MiniProjetAI/Dataset'):
    filename.append(os.path.join('/content/drive/MyDrive/Colab Notebooks/MiniProjetAI/Dataset',file))
```

```python
for file in tqdm(filename):
    feature_list.append(extract_feature(file,model))
```

```python
import pickle

pickle.dump(feature_list,open('/content/drive/MyDrive/Colab Notebooks/MiniProjetAI/featurevector.pkl','wb'))
pickle.dump(filename,open('/content/drive/MyDrive/Colab Notebooks/MiniProjetAI/filenames.pkl','wb'))
```
